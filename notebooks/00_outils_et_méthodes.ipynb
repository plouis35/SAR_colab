## les incertitudes


<font size=5> **1. sur la largeur équivalente (EW) : l'intégrale de la gaussienne ajustée**</font>

$$\sigma_{\text{EW}} \approx \frac{1.5}{\text{SNR}} \sqrt{\delta\lambda \cdot \text{EW}}$$

Elle découle du travail de Cayrel (1988). L'idée est de propager le bruit sur le continuum à travers l'intégrale qui définit l'EW :

$$\text{EW} = \int \left(1 - \frac{F_\lambda}{F_\text{cont}}\right) d\lambda$$

En discrétisant sur $N$ pixels de largeur $\delta\lambda$, et en supposant un bruit gaussien de dispersion $\sigma_\text{pixel} \approx 1/\text{SNR}$ (normalisé au continuum) :

$$\sigma_{\text{EW}} = \sigma_\text{pixel} \cdot \sqrt{N \cdot \delta\lambda^2} = \frac{1}{\text{SNR}}\sqrt{N \cdot \delta\lambda^2}$$

Comme le nombre de pixels est $N \sim \text{EW}/\delta\lambda$, on obtient :

$$\sigma_{\text{EW}} \approx \frac{1}{\text{SNR}}\sqrt{\delta\lambda \cdot \text{EW}}$$

Le facteur **1.5** est empirique et correctif — il tient compte de l'incertitude sur le placement du continuum, qui est la source d'erreur dominante.

<font size=5> **2. sur la vitesse radiale (via le centroïde de la gaussienne ajustée)**</font>


$$\sigma_{v} \approx \frac{c \cdot \text{FWHM}}{2.35 \cdot \text{SNR} \cdot \sqrt{N_{\text{pts}}}}$$

C'est de la statistique standard, et non une formule empirique. C'est l'équivalent spectroscopique de la formule de l'erreur sur la moyenne en statistique :

$$\sigma_{\bar{x}} = \frac{\sigma}{\sqrt{N}}$$

Pour une gaussienne de largeur en écart type retourné par l'ajustement : $$\sigma_g = \text{FWHM}/2.35$$ avec $2.35 \approx \sqrt{8 \ln 2}$, l'incertitude sur son centre est :

$$\sigma_{\lambda_c} = \frac{\sigma_g}{\text{SNR} \cdot \sqrt{N_\text{pts}}} = \frac{\text{FWHM}}{2.35 \cdot \text{SNR} \cdot \sqrt{N_\text{pts}}}$$

On convertit ensuite en vitesse via le décalage Doppler :

$$\sigma_v = c \cdot \frac{\sigma_{\lambda_c}}{\lambda_0} \longrightarrow \frac{c \cdot \text{FWHM}}{2.35 \cdot \text{SNR} \cdot \sqrt{N_\text{pts}}}$$

Le facteur $\sqrt{N_\text{pts}}$ traduit le gain statistique à utiliser le plus de pixels pour contraindre le centre.


<font size=5> **3. sur la largeur à mi-hauteur (FWHM) de la gaussienne ajustée**<font>


la formule simplifiée :

$$ \sigma_{\text{FWHM}} \approx \frac{\text{FWHM}}{\text{SNR}} $$

est couramment utilisée pour de la spectroscopie rapide, mais c'est est une approximation grossière et pessimiste.

Pour être plus précis on reprend l’incertitude sur la largeur $\sigma_g$ (ou FWHM = $2.35 \cdot \sigma_g$) donnée par :

$$ \sigma_{\text{FWHM}} \approx \frac{2.35 \cdot \text{FWHM}}{\text{SNR} \cdot \sqrt{N_{\text{pts}}}} $$

<font size=5>  **on estime les incertitudes systématiques (variations de flexion et erreurs de calibration)** </font>
  1. en prenant 1/10ème de la largeur (FWHM) des raies de la lampe de calibration (présent dans les logs du logiciel specinti par exemple)
  2. et l'erreur RMS issue de la calibration (présent dans les logs du logiciel specinti par exemple)
- et on ajoute quadratiquement ces incertitudes systématiques à l'incertitude statistique vue ci-dessus :

$$ \text{erreur totale} = \sqrt{{\text{erreur statistique}^2} + {\text{erreur neon}^2} + {\text{erreur calibration}^2}} $$


<font size=5> **différences entre Résolution et Précision** </font>
- la *Résolution* (15 km/s), c'est la taille de la pointe de l'outil de mesure
- la *Précision* (1,5 km/s), c'est la capacité à déterminer où se trouve exactement le centre du trait de la pointe de mesure

--> la règle du 1/10ème se justifie : grâce aux ajustements et à un bon SNR, on peut localiser le centre d'une raie avec une précision 10 fois plus fine que la largeur de la raie elle-même

<font size=5> **le SNR détermine la qualité** </font>
- SNR > 100 : mesures précises de centroïdes et vitesses radiales fiables
- SNR 30–100 : FWHM et EW mesurables, vitesses avec incertitude notable
- SNR 10–30 : seules les raies fortes et larges sont exploitables (EW uniquement)
- SNR < 10 : spectre d'identification seulement, aucune mesure métrique fiable


<font size=5> **le chi-carré (χ²) mesure la qualité** </font>
$$χ2 = \sum_{i=1}^{N} \left( \frac{y_i - f(x_i)}{\sigma_i} \right)^2$$
où :
- yi​ = valeur mesurée au pixel i
- f(xi​) = valeur du modèle (la gaussienne fittée)
- σi​ = incertitude sur la mesure (≈ bruit = continuum / SNR)

Chaque terme est le résidu normalisé au carré — si le fit est parfait, chaque résidu vaut environ 1σ, donc chaque terme vaut environ 1.

<font size=5> **Le χ² réduit** </font>

$$χ_v2 = \frac{χ2}{N - p}$$

où p = nombre de paramètres libres du modèle (3 pour une gaussienne : amplitude, centre, largeur).
- ≈ 1.0 --> Fit parfait, le modèle est adapté et le bruit bien estimé
- \>> 1 (ex: 5, 10...) --> Fit mauvais, le modèle inadapté, ou bruit sous-estimé
- \<< 1 (ex: 0.1, 0.2...) --> Sur-ajustement, ou bruit sur-estimé

## Choix du modèle d'ajustement

Pour ajuster un profil, on choisit une fonction mathématique qui modélise l'intensité du flux ($I$) en fonction de la longueur d'onde ($\lambda$) autour de la raie.


| Caractéristiques                      | Gaussienne                                                     | Lorentzienne                              | Voigt                        |
| :------------------------------------ | :------------------------------------------------------------- | :---------------------------------------- | :--------------------------- |
| Ailes                                 | Fines (tombent vite)                                           | Larges (tombent lentement)                | Larges (comme Lorentz)       |
| Zone du Sommet                        | Arrondie                                                       | Pointue                                   | Arrondie (dominée par Gauss) |
| Origine Physique                      | Doppler thermique (l'agitation aléatoire des atomes due à la chaleur) | Amortissement naturel (limite de la durée de vie des niveaux d'énergie) et élargissement par pression   | Combine agitation thermique et effets de pression/collision |
| Utilisation typique                   | VR, faible résolution                                          | Raies dominées par pression (H)           | EW précise, haute résolution |
| Conséquence sur EW (si mauvais choix) | Sous-estimation de l'EW réelle (manque les ailes de pression). | Sur-estimation si le centre est trop rond. | Le meilleur compromis.       |

## un exemple complet pour analyser une raie simulée

````python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
from astropy import units as u
from astropy.modeling import models
from astropy.modeling.fitting import LevMarLSQFitter
from specutils import Spectrum, SpectralRegion
from specutils.fitting import fit_lines
from specutils.analysis import centroid, fwhm, equivalent_width, snr_derived

# =============================================================================
# 1. DONNÉES SIMULATION
# =============================================================================
nb_echantillons = 500
bruit_sigma = 0.04
c_kms = 299792.458
lambda_repos = 6562.80 
lambda_obs = 6564.0
fwhm_neon_pix = 6.0  # pixels (extrait du log de specinti)
sigma_calib = 0.03  # Å (extrait du log de specinti)

# Axe X
x = np.linspace(6550, 6580, nb_echantillons)
delta_lambda = np.mean(np.diff(x))  # Dispersion (Å/pix)

# Profil en ABSORPTION (Gaussienne inversée)
amplitude = 0.5
y = 1.0 - amplitude * np.exp(-0.5 * ((x - lambda_obs)/1.5)**2) 
y += np.random.normal(0, bruit_sigma, len(x)) 

# Création du spectre
spec = Spectrum(flux=y*u.adu, spectral_axis=x*u.AA)

# =============================================================================
# 2. PRÉPARATION DES DONNEES
# =============================================================================
region_raie = SpectralRegion(6555.0 * u.AA, 6570.0 * u.AA)
region_cont = SpectralRegion(6550.0 * u.AA, 6555.0 * u.AA) 

# Calcul Continuum
mask_indices = (x >= region_cont.lower.value) & (x <= region_cont.upper.value)
niveau_continuum = np.mean(spec.flux.value[mask_indices])
snr_val = snr_derived(spec, region_cont).value

# Spectre Normalisé sur le continuum
flux_norm = spec.flux.value / niveau_continuum
spec_norm = Spectrum(flux=flux_norm * u.dimensionless_unscaled, 
                       spectral_axis=spec.spectral_axis)

# Pour absorption : spectre inversé (1 - Flux)
flux_inverse = 1.0 - flux_norm
spec_inverse = Spectrum(flux=flux_inverse * u.dimensionless_unscaled, 
                          spectral_axis=spec.spectral_axis)

# =============================================================================
# 3. PARAMÈTRES INSTRUMENTAUX ET SYSTÉMATIQUES
# =============================================================================
fwhm_inst_angstrom = fwhm_neon_pix * delta_lambda      
sigma_syst_vr = (fwhm_inst_angstrom / 10.0 / lambda_repos) * c_kms  # précision : règle du 1/10
err_vr_calib = (c_kms * sigma_calib / lambda_repos)  

print("=" * 70)
print(" PARAMÈTRES SIMULATION")
print("=" * 70)
print(f"SNR                        : {snr_val:.1f}")
print(f"Lambda repos               : {lambda_repos:.2f}")
print(f"Lambda observée            : {lambda_obs:.2f}")
print(f"Vitesse radiale simulée    : {c_kms*(lambda_obs-lambda_repos)/lambda_repos:.2f} km/s")
print()


print("=" * 70)
print(" PARAMÈTRES INSTRUMENTAUX")
print("=" * 70)
print(f"FWHM Instrumentale (Néon)  : {fwhm_inst_angstrom:.2f} Å")
print(f"Dispersion                 : {delta_lambda:.3f} Å/pix")
print(f"Err. Systématique (Néon)   : {sigma_syst_vr:.2f} km/s")
print(f"Err. Calibration λ         : {sigma_calib:.2f} Å → {err_vr_calib:.2f} km/s")
print()

# =============================================================================
# 4. MESURES SANS AJUSTEMENT AVEC SPECUTILS
# =============================================================================
cent_brut = centroid(spec_inverse, regions=region_raie) 
fwhm_brut = fwhm(spec_inverse, regions=region_raie)     
ew_brut = equivalent_width(spec_norm, regions=region_raie) 

# Incertitudes Statistiques
err_fwhm_stat = fwhm_brut.value / snr_val
N_pts = fwhm_brut.value / delta_lambda
err_cent_stat = fwhm_brut.value / (2.355 * snr_val * np.sqrt(N_pts))
err_ew_stat = (1.5 / snr_val) * np.sqrt(delta_lambda * np.abs(ew_brut.value))

# Vitesse Radiale
vr_brut = c_kms * (cent_brut.value - lambda_repos) / lambda_repos
err_vr_stat = c_kms * err_cent_stat / lambda_repos

# erreur totale avec calibration
err_vr_total = np.sqrt(err_vr_stat**2 + sigma_syst_vr**2 + err_vr_calib**2)

# Vitesse Expansion 
if fwhm_brut.value > fwhm_inst_angstrom:
    fwhm_etoile = np.sqrt(fwhm_brut.value**2 - fwhm_inst_angstrom**2)
    err_fwhm_etoile = (fwhm_brut.value / fwhm_etoile) * err_fwhm_stat
else:
    fwhm_etoile = 0
    err_fwhm_etoile = err_fwhm_stat
    
v_expansion = c_kms * fwhm_etoile / lambda_repos
err_v_expansion = c_kms * err_fwhm_etoile / lambda_repos

print("=" * 70)
print(" RÉSULTATS SANS AJUSTEMENT")
print("=" * 70)
print(f"Centroïde             : {cent_brut.value:.2f} +/- {err_cent_stat:.2f} Å")
#print(f"FWHM observée         : {fwhm_brut.value:.2f} +/- {err_fwhm_stat:.2f} Å")
print(f"FWHM stellaire        : {fwhm_etoile:.2f} +/- {err_fwhm_etoile:.2f} Å")
print(f"Vitesse Radiale       : {vr_brut:.2f} +/- {err_vr_total:.2f} km/s")
print(f"  └─ err Stat: {err_vr_stat:.2f}, err Syst: {sigma_syst_vr:.2f}, err Calib: {err_vr_calib:.2f}")
print(f"Vitesse Expansion     : {v_expansion:.2f} +/- {err_v_expansion:.2f} km/s")
print(f"EW                    : {ew_brut.value:.2f} +/- {err_ew_stat:.2f} Å")
print()

# =============================================================================
# 5. AJUSTEMENT GAUSSIEN
# =============================================================================
# Modèle : Gaussienne en Absorption + Constante
g_init = models.Gaussian1D(amplitude=-0.5, mean=lambda_obs, stddev=1.0) + models.Const1D(1.0)

# Utilisation du fitter pour accéder à la covariance
fitter = LevMarLSQFitter()
g_fit = fitter(g_init, spec_norm.spectral_axis, spec_norm.flux)

# Extraction des paramètres
center_fit = g_fit[0].mean.value
stddev_fit = g_fit[0].stddev.value
fwhm_fit = stddev_fit * 2.35482  # FWHM = (√8 ln 2) * sigma
amp_fit = g_fit[0].amplitude.value

# Extraction des erreurs depuis la matrice de covariance
cov_matrix = fitter.fit_info.get('param_cov', None)

print("=" * 70)
print(" RÉSULTATS AVEC AJUSTEMENT GAUSSIEN")
print("=" * 70)

if cov_matrix is not None and not np.any(np.isnan(cov_matrix)):
    # Indices : [amplitude, mean, stddev, const]
    err_mean_fit = np.sqrt(cov_matrix[1, 1])  # Erreur sur mean
    err_stddev_fit = np.sqrt(cov_matrix[2, 2])  # Erreur sur stddev
    err_amp_fit = np.sqrt(cov_matrix[0, 0])  # Erreur sur amplitude
    
    # Propagation vers FWHM
    err_fwhm_fit = 2.35482 * err_stddev_fit
    
    print("Matrice de covariance : OK")
else:
    # Fallback sur approche SNR
    err_mean_fit = fwhm_fit / (2.35 * snr_val)
    err_fwhm_fit = fwhm_fit / snr_val
    err_stddev_fit = err_fwhm_fit / 2.35482
    err_amp_fit = np.abs(amp_fit) / snr_val
    
    print("Matrice de covariance : NOK → utilisation du SNR")

# Vitesse Radiale
vr_fit = c_kms * (center_fit - lambda_repos) / lambda_repos
err_vr_fit_stat = c_kms * err_mean_fit / lambda_repos
err_vr_fit_total = np.sqrt(err_vr_fit_stat**2 + sigma_syst_vr**2 + err_vr_calib**2)

# Vitesse Expansion 
if fwhm_fit > fwhm_inst_angstrom:
    fwhm_fit_etoile = np.sqrt(fwhm_fit**2 - fwhm_inst_angstrom**2)
    err_fwhm_fit_etoile = (fwhm_fit / fwhm_fit_etoile) * err_fwhm_fit   # on propage le sigma de la fonction quadratique ci-dessus
else:
    fwhm_fit_etoile = 0
    err_fwhm_fit_etoile = err_fwhm_fit

v_exp_fit = c_kms * fwhm_fit_etoile / lambda_repos
err_v_exp_fit = c_kms * err_fwhm_fit_etoile / lambda_repos

# EW en absorption : positive par convention
# Pour une gaussienne : EW = sqrt(2π) * |amplitude| * sigma
# (intégrale de la gaussienne normalisée au continuum)
ew_fit_val = np.sqrt(2 * np.pi) * np.abs(amp_fit) * stddev_fit

# Propagation : δEW = EW * sqrt[(δA/A)² + (δσ/σ)²]
if amp_fit != 0 and stddev_fit != 0:
    err_ew_fit = ew_fit_val * np.sqrt(
        (err_amp_fit / np.abs(amp_fit))**2 + 
        (err_stddev_fit / stddev_fit)**2
    )
else:
    err_ew_fit = np.abs(ew_fit_val) / snr_val

print(f"Centroïde             : {center_fit:.2f} +/- {err_mean_fit:.2f} Å")
#print(f"FWHM observée         : {fwhm_fit:.2f} +/- {err_fwhm_fit:.2f} Å")
print(f"FWHM stellaire        : {fwhm_fit_etoile:.2f} +/- {err_fwhm_fit_etoile:.2f} Å")
print(f"Vitesse Radiale       : {vr_fit:.2f} +/- {err_vr_fit_total:.2f} km/s")
print(f"  └─ err Stat: {err_vr_fit_stat:.2f}, err Syst: {sigma_syst_vr:.2f}, err Calib: {err_vr_calib:.2f}")
print(f"Vitesse Expansion     : {v_exp_fit:.2f} +/- {err_v_exp_fit:.2f} km/s")
print(f"EW                    : {ew_fit_val:.2f} +/- {err_ew_fit:.2f} Å")

# =============================================================================
# 6. QUALITE DES DONNEES
# =============================================================================

# Calcul du chi-deux réduit
residus = spec_norm.flux.value - g_fit(spec_norm.spectral_axis)
sigma_bruit = 1.0 / snr_val  # bruit estimé depuis le SNR
mask_raie = (x >= region_raie.lower.value) & (x <= region_raie.upper.value)
chi2 = np.sum((residus[mask_raie] / sigma_bruit)**2)
n_points = np.sum(mask_raie)
n_params = 4  # amplitude, mean, stddev, constante
chi2_reduit = chi2 / (n_points - n_params)

print(f"χ² réduit             : {chi2_reduit:.2f}")
print()
#if chi2_reduit < 0.5:
#    print("⚠ Bruit surestimé ou sur-ajustement")
#elif chi2_reduit > 2.0:
#    print("⚠ Fit insuffisant — modèle à revoir ou raie trop complexe")
#else:
#    print("✓ Fit acceptable")
    

# =============================================================================
# 7. VISUALISATION
# =============================================================================
fig = plt.figure(figsize=(8, 6))
gs = gridspec.GridSpec(2, 1, height_ratios=[3, 1]) 
ax1 = fig.add_subplot(gs[0])  # graphique du spectre ajusté
ax2 = fig.add_subplot(gs[1])  # graphique des résidus


ax1.plot(x, spec_norm.flux.value, 'k-', alpha=0.6, label='Brut')
ax1.plot(x, g_fit(x*u.AA), 'r--', linewidth=2, label='Fit Gaussien')
ax1.axvline(lambda_repos, color='blue', linestyle=':', alpha=0.5, label='Hα repos')
ax1.axvline(center_fit, color='red', linestyle=':', alpha=0.5, label='Centroïde fit')
ax1.set_ylabel('Flux')
ax2.set_xlabel('Longueur d\'onde (Å)', fontsize=12)
ax1.set_title('Spectre Hα simulé', fontsize=14, fontweight='bold')
ax1.legend(loc='best')
ax1.grid(True, alpha=0.3)

# on récupère les résidus en divisant le brut par le fit
residus = spec_norm.flux.value - g_fit(x*u.AA)

ax2.scatter(spec_norm.spectral_axis, residus, s=10, color='gray', alpha=0.5)
ax2.axhline(0, color='red', linestyle='--')
ax2.set_ylabel('Résidus')
ax2.set_xticklabels([])  
ax2.grid(alpha=0.3)

# on vérifie que la zone de la raie est 'plate' dans la région de la raie -> le fit encadre bien la raie



plt.tight_layout()
plt.show()


````

## monte carlo

- En spectroscopie, quand on mesure une quantité (comme la largeur équivalente d'une raie $H\alpha$, ou la position d'un pic pour la vitesse radiale), on utilise un spectre unique.
- avec 1000 spectres de la même étoile, on aurait eu 1000 valeurs légèrement différentes à cause du bruit de poisson et de lecture.
- La méthode Monte Carlo simule cette réalité : elle prend le spectre "unique", génère $N$ copies virtuelles, ajoute du bruit aléatoire supplémentaire basé sur l'incertitude des données.
- On refait la mesure sur ces 1000 spectres. L'écart-type de ces 1000 mesures donne la **barre d'erreur officielle**.
- C'est plus robuste que les formules analytiques théoriques, surtout si le spectre est complexe ou asymétrique.


````python

def monte_carlo_simulation(flux_original, axis_x, bruit_sigma, n_iter=1000):
    """
    Lance N simulations en ajoutant du bruit aléatoire et mesure la dispersion.
    """    
    print(f"\n\rLancement Monte Carlo ({n_iter} itérations)...")
    
    # Tableaux pour stocker les résultats
    res_rv = []
    res_fwhm = []
    res_ew = []
    n_echecs = 0
    
    # On initialise large pour aider le fit
    #g_init_mc = models.Gaussian1D(amplitude=-0.5, mean=6564, stddev=1.0) + models.Const1D(1.0)
    # Initialisation depuis le résultat du fit principal
    g_init_mc = models.Gaussian1D(
        amplitude=amp_fit, 
        mean=center_fit, 
        stddev=stddev_fit
    ) + models.Const1D(g_fit[1].amplitude.value)
    
    for i in range(n_iter):
        # 1. Génération du bruit
        # On ajoute un bruit aléatoire proportionnel au sigma mesuré
        noise = np.random.normal(0, bruit_sigma, len(flux_original))
        flux_noisy = flux_original + noise
        
        # Création du spectre temporaire
        spec_mc = Spectrum(flux=flux_noisy*u.dimensionless_unscaled, spectral_axis=axis_x)
        
        try:
            # 2. Mesures (On reprend la  méthode "Avec Ajustement")
            g_fit_mc = fit_lines(spec_mc, g_init_mc)
            
            # Récupération des valeurs
            center = g_fit_mc[0].mean.value
            fwhm = g_fit_mc[0].fwhm.value
            
            # Calcul EW (Approximation gaussienne rapide et robuste)
            amp = g_fit_mc[0].amplitude.value
            std = g_fit_mc[0].stddev.value
            ew = (np.sqrt(2*np.pi) * np.abs(amp) * std)
            
            # Stockage
            res_rv.append(center)
            res_fwhm.append(fwhm)
            res_ew.append(ew)
            
        except Exception:
            n_echecs += 1

    arr_rv = np.array(res_rv)
    arr_fwhm = np.array(res_fwhm)
    arr_ew = np.array(res_ew)
    
    # Conversion RV en km/s (pour l'écart-type)
    c_kms = 299792.458
    lambda_ref = 6562.80
    # L'écart-type sur la position (Angstroms) converti en vitesse
    sigma_stat_rv = (np.std(arr_rv) / lambda_ref) * c_kms

    taux_echec = n_echecs / n_iter * 100
    if taux_echec > 5:
        print(f"⚠ Taux d'échec : {taux_echec:.1f}% — vérifier les conditions initiales du fit")
    else:
        print(f"✓ Taux d'échec : {taux_echec:.1f}%")

    fig, axes = plt.subplots(1, 3, figsize=(12, 4))

    # on affiche la distribution des valeurs
    for ax, data, label, unit, analytical in zip(
        axes,
        [(arr_rv - lambda_repos) * c_kms / lambda_repos, arr_fwhm, arr_ew],  # données MC
        ['Vitesse radiale', 'FWHM', 'EW'],
        ['km/s', 'Å', 'Å'],
        [sigma_stat_rv, np.std(arr_fwhm), np.std(arr_ew)]
        ):
            ax.hist(data, bins=50, density=True, alpha=0.6, color='steelblue', label='MC')
            
            # Gaussienne analytique superposée
            mu, sigma = np.mean(data), np.std(data)
            x_g = np.linspace(mu - 4*sigma, mu + 4*sigma, 200)
            ax.plot(x_g, np.exp(-0.5*((x_g-mu)/sigma)**2) / (sigma*np.sqrt(2*np.pi)), 
                    'r-', lw=2, label=f'σ={sigma:.3f} {unit}')
            
            ax.axvline(mu, color='red', linestyle='--', alpha=0.7)
            ax.set_xlabel(f'{label} ({unit})')
            ax.set_ylabel('Densité')
            ax.legend(fontsize=8)
            ax.grid(True, alpha=0.3)
    
    plt.suptitle(f"Distribution Monte Carlo — {n_iter} itérations", fontweight='bold')
    plt.tight_layout()
    plt.show()
    

    return {
        "sigma_rv": sigma_stat_rv,      # Erreur Statistique RV (km/s)
        "sigma_fwhm": np.std(arr_fwhm), # Erreur Statistique FWHM (Angstroms)
        "sigma_ew": np.std(arr_ew)      # Erreur Statistique EW (Angstroms)
    }

    
# 1. On utilise le SNR mesuré comme en situation réelle
niveau_bruit_mc = np.median(spec_norm.flux.value[mask_indices]) / snr_val

# 2. On lance la simulation
mc_results = monte_carlo_simulation(spec_norm.flux.value, spec_norm.spectral_axis, 
                                    bruit_sigma=niveau_bruit_mc, n_iter=1000)

print("-" * 30 + " RÉSULTATS MONTE CARLO " + "-" * 30)
print(f"Incertitude VR   : +/- {mc_results['sigma_rv']:.3f} km/s")
print(f"Incertitude FWHM : +/- {mc_results['sigma_fwhm']:.3f} Å")
print(f"Incertitude EW   : +/- {mc_results['sigma_ew']:.3f} Å")


````
