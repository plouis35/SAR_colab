# analyse : que peut-on mesurer ?

## les méthodes utilisée par les amateurs

| Mesure | Méthode | Implémentation | Incertitude |
| :--- | :--- | :--- | :--- |
| **Qualité des données** | **déviation standard** du flux ($\sigma_{flux}$) dans une région de continuum considérée comme "pure bruit" : $$\text{SNR} = \frac{1}{\sigma_{flux}}$$ |  `1 / astropy.stats.mad_std` | - |
| **Type spectral ou température**| **Ajustement** de la courbe de Planck : $$B_\lambda(T) = \frac{2hc^2}{\lambda^5} \cdot \frac{1}{e^{\frac{hc}{\lambda k_B T}} - 1}$$ | `popt, pcov = scipy.curve_fit` | $$\sigma_T = \sqrt{\text{pcov}[0][0]}$$ |
| **Vitesse de Rotation** ou **Élargissement Thermique** |**Largeur à mi-hauteur (FWHM)** d'un profil normalisé et ajusté. |  `specutils.fitting.fit_lines` | $\sigma_{\text{FWHM}} \approx \frac{\text{2 * FWHM}}{\text{SNR}}$ |
| **Abondance Chimique** | **EW** (Largeur Équivalente) : **aire** d'un profil normalisé et ajusté. |  `specutils.analysis.equivalent_width` | $$\sigma_{\text{EW}} \approx \frac{1.5}{\text{SNR}} \sqrt{\delta\lambda \cdot \text{EW}}$$ |
| **Vitesse de l'Étoile** (Déplacement global de l'objet) | **Centroïde** (Mono-raie) : Le **centre** ($x_0$) ajusté d'un profil nomalisé et ajusté. | `specutils.fitting.fit_lines` | $$\sigma_{v} \approx \frac{c \cdot \text{FWHM}}{2.35 \cdot \text{SNR} \cdot \sqrt{N_{\text{pts}}}} $$|
| **Mouvement orbital** (de binaires) | **Centroïde** (Multi-raies / CCF) : **moyenne** des vitesses radiales obtenues sur plusieurs raies individuelles. |  `specutils.correlation.template_correlate` | $$\sigma_{\text{tot}} = \frac{\text{StdDev}(\text{Liste VR})}{\sqrt{N_{\text{raies}}}}$$ |


**notes sur les incertitudes**
- en spectro amateur HR (starEx et LiRES), on peut simplifier les incertitudes à partir de la largeur des raies de la lampe de calibration (néon)
- cette méthode est cohérente avec la résolution ($R$) : Avec $R \approx 20\,000$, le "pixel de résolution" est d'environ $15 \text{ km/s}$ ($\Delta v \approx \frac{c}{R}$)
- on peut en théorie mesurer à $1.5 \text{ km/s}$ près (soit $1/10^{ème}$ de la résolution), c'est déjà être extrêmement ambitieux et performant !

**nuance entre Résolution et Précision**
- la *Résolution* (15 km/s), c'est la taille de la pointe de l'outil de mesure
- la *Précision* (1,5 km/s), c'est la capacité à déterminer où se trouve exactement le centre du trait de la pointe de mesure
- --> la règle du 1/10ème se justifie : grâce aux ajustements et à un bon SNR, on peut localiser le centre d'une raie avec une précision 10 fois plus fine que la largeur de la raie elle-même

**le SNR détermine la qualité**
- si SNR > 100 : le bruit influe peu, on fait confiance aux mathématiques associées à la gaussienne ajustée
- si SNR < 100 : l'instrument limite la précision (flexions, taille des pixels, calibration)
- si SNR < 10, vous ne pouvez plus rien mesurer du tout, l'erreur est de l'ordre de la raie elle-même


Pour ajuster un profil, on utilise  3 modèles. Chacun est une fonction mathématique qui modélise l'intensité du flux ($I$) en fonction de la longueur d'onde ($\lambda$) autour de la raie.


| Caractéristiques                      | Gaussienne                                                     | Lorentzienne                              | Voigt                        |
| :------------------------------------ | :------------------------------------------------------------- | :---------------------------------------- | :--------------------------- |
| Ailes                                 | Fines (tombent vite)                                           | Larges (tombent lentement)                | Larges (comme Lorentz)       |
| Zone du Sommet                        | Arrondie                                                       | Pointue                                   | Arrondie (dominée par Gauss) |
| Origine Physique                      | Doppler thermique (l'agitation aléatoire des atomes due à la chaleur) | Amortissement naturel (limite de la durée de vie des niveaux d'énergie) et élargissement par pression   | Combine agitation thermique et effets de pression/collision |
| Utilisation typique                   | VR, faible résolution                                          | Raies dominées par pression (H)           | EW précise, haute résolution |
| Conséquence sur EW (si mauvais choix) | Sous-estimation de l'EW réelle (manque les ailes de pression). | Sur-estimation si le centre est trop rond. | Le meilleur compromis.       |

## une démonstration complète

````python
import numpy as np
import matplotlib.pyplot as plt
from astropy import units as u
from astropy.modeling import models
from specutils import Spectrum, SpectralRegion
from specutils.fitting import fit_lines
from specutils.analysis import centroid, fwhm, equivalent_width, snr, snr_derived

# --- 1. données pour la simulation ---
nb_echantillons = 500
cont_reel = 1.0
bruit_sigma = 0.1
raie_obs = 6563.44
amplitude = 15
c_kms = 299792.458
lambda_repos = 6562.80 

# Raie gaussienne centrée sur raie_obs
x = np.linspace(6400, 6700, nb_echantillons)
y = cont_reel + amplitude * np.exp(-0.5 * ((x - raie_obs)/5)**2) + np.random.normal(0, bruit_sigma, nb_echantillons)
spec = Spectrum(flux=y*u.adu, spectral_axis=x*u.AA)

# --- 2. mesures SANS ajustement ---
region_raie = SpectralRegion(6540.0 * u.AA, 6590.0 * u.AA)
region_cont = SpectralRegion(6400.0 * u.AA, 6500.0 * u.AA)
mask_indices = (x >= region_cont.lower.value) & (x <= region_cont.upper.value)

# On calcule la moyenne du flux dans ces indices
niveau_continuum_brut = np.mean(spec.flux[mask_indices])

# Normalisation : on divise par le niveau du continuum
flux_norm_val = spec.flux.value / niveau_continuum_brut.value
spec_norm = Spectrum(flux=flux_norm_val * u.dimensionless_unscaled, 
                       spectral_axis=spec.spectral_axis)

# on garde une version sans continuum (à 0) pour le centroïde et la FWHM
spec_norm_sans_cont = spec_norm - models.Const1D(1.0)(spec.spectral_axis)

# Mesures
snr_val = snr_derived(spec, region_cont)
cent_brut = centroid(spec_norm_sans_cont, regions=region_raie)
vr_brut = c_kms * (cent_brut.value - lambda_repos) / lambda_repos
fwhm_brut = fwhm(spec_norm_sans_cont, regions=region_raie)
ew_brut = equivalent_width(spec_norm, regions=region_raie)

# Incertitudes - méthode de Cayrel
delta_lambda = np.mean(np.diff(x))       # pas spectral (dispersion)
n_pts_brut = fwhm_brut.value / delta_lambda

err_fwhm_brut = (2.0 * fwhm_brut.value) / snr_val.value   
err_ew_brut = (1.5 / snr_val.value) * np.sqrt(delta_lambda * np.abs(ew_brut))

err_cent_brut = fwhm_brut.value / (2.35 * snr_val.value * np.sqrt(n_pts_brut)) 
err_vr_brut = (c_kms * err_cent_brut) / lambda_repos

v_expansion_brut = (c_kms * (fwhm_brut.value / lambda_repos)) / 2
err_v_exp_brut = (c_kms * err_fwhm_brut) / (2 * lambda_repos)

print(f"SNR mesuré : {snr_val:.0f}")
print("-" * 30 + " SANS AJUSTEMENT " + "-" * 30)
print(f"Centroïde : {cent_brut.value:.2f} +/-{err_cent_brut:.2f} Å")
print(f"FWHM      : {fwhm_brut.value:.2f} +/-{err_fwhm_brut:.2f} Å")
print(f"EW        : {ew_brut.value:.2f} +/-{err_ew_brut.value:.2f} Å")
print(f"Vitesse Radiale : {vr_brut:.2f} +/-{err_vr_brut:.2f} km/s")
print(f"Vitesse d'Expansion : {v_expansion_brut:.2f} +/-{err_v_exp_brut:.2f} km/s")

# --- 3. mesures AVEC ajustement ---
# On définit le modèle : Gaussienne + Continuum (Const1D)
g_init = models.Gaussian1D(amplitude=amplitude*u.adu, mean=6563.4*u.AA, stddev=5*u.AA) + models.Const1D(1.0)
g_fit = fit_lines(spec, g_init)

# on extract les résultats (le [0] est la gaussienne)
center_fit = g_fit[0].mean.value
fwhm_fit = g_fit[0].fwhm.value

# Incertitudes
delta_lambda = np.mean(np.diff(x))       # pas spectral (dispersion)
n_pts = fwhm_fit / delta_lambda

err_fwhm_fit = (2.0 * fwhm_fit) / snr_val.value
err_cent_fit = fwhm_fit / (2.35 * snr_val.value * np.sqrt(n_pts))

print("-" * 30 + " AVEC AJUSTEMENT " + "-" * 30)
print(f"Centroïde : {center_fit:.2f} +/-{err_cent_fit:.2f} Å")
print(f"FWHM      : {fwhm_fit:.2f} +/-{err_fwhm_fit:.2f} Å")

# Extraction des paramètres pour le calcul de l'aire
amp_fit = g_fit[0].amplitude.value
sigma_fit = g_fit[0].stddev.value
continuum_fit = g_fit[1].amplitude.value

# Calcul de l'EW du modèle
# L'aire d'une gaussienne est Amp * Sigma * sqrt(2*pi)
# L'EW est cette aire divisée par le niveau du continuum
aire_gaussienne = amp_fit * sigma_fit * np.sqrt(2 * np.pi)
ew_fit_val = aire_gaussienne / continuum_fit

# vitesse radiale
vr_fit = c_kms * (center_fit - lambda_repos) / lambda_repos
err_vr_fit = (c_kms * err_cent_fit) / lambda_repos

v_expansion = (c_kms * (fwhm_fit / lambda_repos)) / 2
err_v_exp = (c_kms * err_fwhm_fit) / (2 * lambda_repos)

# Incertitude sur l'EW du fit (Formule de Cayrel)
# On peut diminuer le facteur 1.5 plus bas car le fit réduit l'incertitude de 30% (cf. Landman et al.)
err_ew_fit = (1.5 / snr_val.value) * np.sqrt(delta_lambda * np.abs(ew_fit_val))
print(f"EW        : {ew_fit_val:.2f} +/-{err_ew_fit:.2f} Å")
print(f"Vitesse Radiale : {vr_fit:.2f} +/-{err_vr_fit:.2f} km/s")
print(f"Vitesse d'Expansion : {v_expansion:.2f} +/-{err_v_exp:.2f} km/s")
print("")

# on affiche le spectre et son ajustement
y_fit = g_fit(spec.spectral_axis)
plt.close('all')
plt.plot(x, y, label="brut", alpha=1.0)
plt.plot(x, y_fit, label="ajusté", ls='--')
plt.title('raie simulée')
plt.grid(True)
plt.legend()
plt.show()

````

````python
# Simuler un spectre H-alpha (autour de 6563 Angstroms) avec bruit
x = np.linspace(6400, 6700, 500)

dash.clear_spectra()

# Profil gaussien centré sur H-alpha
y1 = 10 + 50 * np.exp(-0.5 * ((x - 6562.8)/5)**2)  + np.random.normal(0, 1, 500)
dash.show_spectrum(x, y1, label="Centre")

# Simuler un deuxième spectre décalé (effet Doppler)
y2 = 8 + 40 * np.exp(-0.5 * ((x - 6570)/5)**2) + np.random.normal(0, 1, 500)
dash.show_spectrum(x, y2, label="Bras Spiral (+300km/s)")
````

````python
from astropy import units as u
from astropy.nddata import CCDData
import ccdproc as ccdproc

# lecture d'un FIT 2D
img = CCDData.read('data/20250807_deneb_ruchbah_gamcas/_deneb_20250807_902_2D.fit', unit=u.Unit('adu'))
dash.show_image(img, img.meta['OBJECT'])
````

````python
from specutils import Spectrum

dash.clear_spectra()

# lecture de fichiers DAT (csv)
spec_name = 'data/MilesLibraryBASS/G0V s0176 HD034411.dat'
spec_name = 'data/MilesLibraryBASS/G0 s0034 HD004906.dat'


_spc_array = np.loadtxt(spec_name)
_spec1d = Spectrum(spectral_axis=_spc_array[:,0] * u.Unit('Angstrom'), 
                 flux=_spc_array[:,1] * u.Unit('mJy')
                         )
    
dash.show_spectrum(_spec1d.wavelength, _spec1d.flux, label=spec_name, ls='--', color='green')
````
