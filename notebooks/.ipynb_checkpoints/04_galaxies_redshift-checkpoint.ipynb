# chargement du spectre

```python
# imports
import numpy as np
import astropy.units as u
from specutils import Spectrum1D
from astropy.modeling.models import Gaussian1D
from astropy.modeling.fitting import LevMarLSQFitter # L'outil de fit correct

from specutils.fitting import fit_generic_continuum
from astropy.modeling.models import Polynomial1D # Exemple de modèle de fit
from specutils import SpectralRegion

from astropy.constants import c
from astropy.cosmology import Planck18

# on affiche le spectre de la galaxie

db.clear_spectra()

# lecture de fichiers DAT (csv)
#spec_name = 'data/plouis/_ngc7215-_20161228_896_P.Louis.dat'
#spec_name = 'data/plouis/_ngc4151_20170121_974_P.Louis.dat'
spec_name = 'data/plouis/_3c273_20170407_870_PLouis.dat'

_spc_array = np.loadtxt(spec_name)
_spec1d = Spectrum1D(spectral_axis=_spc_array[:,0] * u.Unit('Angstrom'), 
                 flux=_spc_array[:,1] * u.Unit('mJy')
                         )
    
db.load_spectrum(_spec1d.wavelength, _spec1d.flux, label=spec_name)
```

# retrait du continuum

```python

# on reprend le spectre créé ci-dessus
spectrum = _spec1d #Spectrum1D(flux=flux, spectral_axis=wavelength)

continuum_model_fit = Polynomial1D(degree=3) 

# on défini les zones à exclure
windows_tuples = [
    (4850 * u.Angstrom, 4900 * u.Angstrom), # Exclusion de Hβ
    (4950 * u.Angstrom, 5040 * u.Angstrom)  # Exclusion des [OIII]
]

# CRÉATION DE LA LISTE D'OBJETS SPECTRALREGION :
exclusion_regions = [
    SpectralRegion(start, end)
    for start, end in windows_tuples
    ]

# on ajuste le modèle uniquement aux régions qui NE SONT PAS dans exclusion_windows.
fitted_continuum = fit_generic_continuum(
    spectrum,
    model=continuum_model_fit,
    exclude_regions=exclusion_regions
)

# Évaluation du modèle ajusté sur l'intégralité du spectre
continuum_flux_to_subtract = fitted_continuum(spectrum.spectral_axis)

# Obtention du spectre net
spectrum_sans_cont = Spectrum1D(
    flux=spectrum.flux - continuum_flux_to_subtract,
    spectral_axis=spectrum.spectral_axis
)

db.load_spectrum(spectrum.spectral_axis.to(u.Angstrom).value, continuum_flux_to_subtract.value, label='continuum')
db.load_spectrum(spectrum_sans_cont.spectral_axis, spectrum_sans_cont.flux, label='continuum retiré')
```

# ajustement 

```python
db.clear_spectra()

# on recharge le spectre sans le continnum
db.load_spectrum(spectrum_sans_cont.spectral_axis, spectrum_sans_cont.flux, label='continuum retiré')

# Initialiser l'objet Fitter (Levenberg-Marquardt)
fitter = LevMarLSQFitter()

# on va prendre OIII
#lambda_rest_oiii = 5006.84 * u.Angstrom 
lambda_rest_oiii = 4861.0 * u.Angstrom 


# on défini le modèle initial (Gaussien simple)
initial_model = Gaussian1D(
    amplitude=.02 * u.Jy,
    mean=5640 * u.Angstrom, #5086.84 * u.Angstrom,
    stddev=5.0 * u.Angstrom,
    
    # bornes pour la moyenne ou l'amplitude (double pics ou trop rapprochés)
    #bounds={
     #   'mean': (5050 * u.Angstrom, 5400 * u.Angstrom),
     #   # 'amplitude' : () 
    #}
)

# on fait l'ajustement
fitted_model = fitter(initial_model, spectrum_sans_cont.spectral_axis, spectrum_sans_cont.flux)

# on affiche le fit
db.load_spectrum(
    spectrum_sans_cont.spectral_axis.to(u.Angstrom).value,
    fitted_model(spectrum_sans_cont.spectral_axis).value,
    label='Modèle Gaussien ajusté'
)

l_OIII = db.ax_spec.axvline(lambda_rest_oiii.value, color='red', ls='--', alpha=1, lw=1)

```

# z et distance
Redshift ($z$)$$z = \frac{\lambda_{obs} - \lambda_{rest}}{\lambda_{rest}}$$

```python

# Conversion avec unités
lambda_obs_fit = fitted_model.mean * fitted_model.mean.unit
amplitude_fit = fitted_model.amplitude * fitted_model.amplitude.unit
stddev_fit = fitted_model.stddev * fitted_model.stddev.unit

# résultats
print(f"--- Résultat de l'ajustement autour de {lambda_rest_oiii} ---")
print(f"Amplitude ajustée : {amplitude_fit.round(2)}")
print(f"Centroïde (λ_obs_fit) : {lambda_obs_fit.round(3)}")
print(f"Écart-type : {stddev_fit.round(2)}")

lambda_obs_oiii = fitted_model.mean # pour le print après

z_oiii = (lambda_obs_oiii - lambda_rest_oiii) / lambda_rest_oiii
print(f"Redshift (z) basé sur [OIII] : {z_oiii.decompose().round(4)}")

# Extraction de la valeur numérique de z
z_value = z_oiii.decompose().value 

# Vr = c * z
vitesse_radiale = z_value * c.to(u.km/u.s)

print(f"Vitesse Radiale (Vr) corrigée : {vitesse_radiale.round(2)}")

distance_mpc_oiii = Planck18.luminosity_distance(z_oiii)
distance_pc_oiii = distance_mpc_oiii.to(u.pc)

print(f"Distance (D_L) en Mpc : {distance_mpc_oiii.round(2)}")

```
