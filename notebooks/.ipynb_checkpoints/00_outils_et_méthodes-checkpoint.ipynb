# ipywidgets : ajouter de l'interactivité à votre code

````python
import ipywidgets as widgets
from IPython.display import display

# creation et affichage de 'widgets' (éléments de dialogues d'interface)
display(slider_widget := widgets.IntSlider(value=50, min=0, max=100, step=1, description='Slider:', orientation='horizontal'))
display(text_widget := widgets.Text(value='Enter your name', description='Name:'))
display(dropdown_widget := widgets.Dropdown(options=['Option 1', 'Option 2', 'Option 3'], value='Option 1', description='Dropdown:'))
display(checkbox_widget := widgets.Checkbox(value=False, description='Checkbox:'))
display(button_widget := widgets.Button(description='Click Me', button_style='info', tooltip='Click me', icon='eraser'))
display(toggle_button_widget := widgets.ToggleButton(value=False,description='Toggle:', button_style='danger', tooltip='Toggle me', icon='toggle-off'))
display(color_picker_widget := widgets.ColorPicker(value='#ff0000', description='Color:'))
display(date_picker_widget := widgets.DatePicker(description='Pick a Date:'))
display(my_pb := widgets.IntProgress(value=4,min=0,max=10,description='Loading:', bar_style= 'success',  style={'bar_color': 'green'}, orientation='horizontal'))

# actions (callback)
def print_output(change):
    with output_cell:
        output_cell.clear_output()
        print(f"change dict = {change}") #['new']}")

# association des actions
toggle_button_widget.observe(print_output, names='value')
date_picker_widget.observe(print_output, names='value')
checkbox_widget.observe(print_output, names='value')
dropdown_widget.observe(print_output, names='value')
text_widget.observe(print_output, names='value')
slider_widget.observe(print_output, names='value')
color_picker_widget.observe(print_output, names='value')
button_widget.on_click(print_output)

# affichage dans une cellule dédiée
output_cell = widgets.Output()
display (output_cell)
````

````python
import pathlib

# le combiner gère une liste de fichiers 'triable' (par date, type, répertoire...)
img_list = [CCDData.read(f, unit=u.Unit('adu')) 
            for f in list(pathlib.Path('data/20250807_deneb_ruchbah_gamcas/').glob('den*.fit'))]

print(f"{len(img_list)} images à combiner")

# on 'combine' les images (sum, median, average, sigma_clip, trim, etc...)
img_combined = ccdproc.Combiner(img_list).sum_combine() #.median_combine(), ...
print(img_combined)
````

````python
# on calcule les incertitudes avec propagation auto 
ccdproc.create_deviation(img_combined, gain=1.5 * u.electron/u.adu, readnoise=5 * u.electron)
img_combined.uncertainty
````

## specutils : outils d'analyse spectrale
https://specutils.readthedocs.io/en/latest/index.html#

````python
from specutils import Spectrum1D
from specutils.fitting import fit_generic_continuum, fit_continuum
from specutils.analysis import centroid, fwhm, snr, snr_derived, equivalent_width, line_flux, gaussian_fwhm
from specutils.fitting import estimate_line_parameters
from specutils.spectra import SpectralRegion
from specutils.manipulation import extract_region
from specutils.fitting import fit_lines


# exemples d'utilisation dans les sections d'analyse de cas
````

````python
import numpy as np
import matplotlib.pyplot as plt
from astropy.modeling import models
from astropy import units as u

plt.close('all')

# gaussienne simulée
np.random.seed(0)
x = np.linspace(0., 10., 200)
y = 3 * np.exp(-0.5 * (x- 6.3)**2 / 0.8**2)
y += np.random.normal(0., 0.2, x.shape)
spec = Spectrum1D(flux=y*u.adu, spectral_axis=x*u.AA)

print(f"{snr_derived(spec, SpectralRegion(1.0 * u.AA, 3.0 * u.AA))=}")

print(30*'-' +'infos SANS ajustement')
spec_with_continuum = spec + 1 * u.adu
print(f"{centroid(spec_with_continuum, SpectralRegion(4.0 * u.AA, 8.0 * u.AA))=}")
print(f"{equivalent_width(spec_with_continuum)=}")
print(f"{fwhm(spec_with_continuum)=}")

# infos AVEC ajustement
print(30*'-' +'infos AVEC ajustement')
g_init = models.Gaussian1D(amplitude=3.*u.adu, mean=6.1*u.AA, stddev=1.*u.AA)
g_fit = fit_lines(spec, g_init)
y_fit = g_fit(x*u.AA)
g_spec = Spectrum1D(flux=y_fit*u.adu, spectral_axis=x*u.AA)
g_spec_with_continuum = g_spec + 1 #* u.adu
print(f"{centroid(g_spec_with_continuum, SpectralRegion(4.0 * u.AA, 8.0 * u.AA))=}")
print(f"{equivalent_width(g_spec_with_continuum)=}")
print(f"{fwhm(g_spec_with_continuum)=}")

print(30*'-')

# Plot the original spectrum and the fitted.
plt.plot(x, y, label="brut")
plt.plot(x, y_fit, label="ajusté")
plt.title('raie simulée')
plt.grid(True)
plt.legend()
plt.show()


````

# analyse : que mesure-t-on ?

## les méthodes utilisée par les amateurs

| Mesure | Méthode | Implémentation | Incertitude |
| :--- | :--- | :--- | :--- |
| **Qualité des données** | **déviation standard** du flux ($\sigma_{flux}$) dans une région de continuum considérée comme "pure bruit" : $$\text{SNR} = \frac{1}{\sigma_{flux}}$$ |  `1 / astropy.stats.mad_std` | - |
| **Type spectral ou température**| **Ajustement** de la courbe de Planck : $$B_\lambda(T) = \frac{2hc^2}{\lambda^5} \cdot \frac{1}{e^{\frac{hc}{\lambda k_B T}} - 1}$$ | `popt, pcov = scipy.curve_fit` | $$\sigma_T = \sqrt{\text{pcov}[0][0]}$$ |
| **Vitesse de Rotation** ou **Élargissement Thermique** |**Largeur à mi-hauteur (FWHM)** d'un profil normalisé et ajusté. |  `specutils.fitting.fit_lines` ou ` astropy (LevMarLSQFitter)` | $\sigma_{\text{FWHM}} \approx \frac{\text{FWHM}}{\text{SNR}}$ |
| **Abondance Chimique** (Concentration des éléments) | **EW** (Largeur Équivalente) : **aire** d'un profil normalisée et ajustée. |  `specutils.analysis.equivalent_width` | $$\sigma_{\text{EW}} \approx \frac{1.5}{\text{SNR}} \sqrt{\delta\lambda \cdot \text{EW}}$$ |
| **Vitesse de l'Étoile** (Déplacement global de l'objet) | **VR / Centroïde** (Mono-raie) : Le **centre** ($x_0$) ajusté d'un profil nomalisé et ajusté. | `specutils.fitting.fit_lines`  ou ` astropy (LevMarLSQFitter)` | $$\sigma_{v} \approx \frac{c \cdot \text{FWHM}}{2.35 \cdot \text{SNR} \cdot \sqrt{N_{\text{pts}}}}$$ |
| **Mouvement orbital** (de binaires) | **VR Global** (Multi-raies / CCF) : **moyenne** des vitesses radiales obtenues sur plusieurs raies individuelles. |  `corr, lag = specutils.correlation.template_correlate(spec_observed, spec_template)` | $$\sigma_{\text{tot}} = \frac{\text{StdDev}(\text{Liste VR})}{\sqrt{N_{\text{raies}}}}$$ |


Voici les trois principaux modèles utilisés en spectroscopie. Chacun de ces profils est une fonction mathématique qui modélise l'intensité du flux ($I$) en fonction de la longueur d'onde ($\lambda$) autour de la raie.


| Caractéristiques                      | Gaussienne                                                     | Lorentzienne                              | Voigt                        |
| :------------------------------------ | :------------------------------------------------------------- | :---------------------------------------- | :--------------------------- |
| Ailes                                 | Fines (tombent vite)                                           | Larges (tombent lentement)                | Larges (comme Lorentz)       |
| Zone du Sommet                        | Arrondie                                                       | Pointue                                   | Arrondie (dominée par Gauss) |
| Origine Physique                      | Doppler thermique (l'agitation aléatoire des atomes due à la chaleur) | Amortissement naturel (limite de la durée de vie des niveaux d'énergie) et élargissement par pression   | Combine agitation thermique et effets de pression/collision |
| Utilisation typique                   | VR, faible résolution                                          | Raies dominées par pression (H)           | EW précise, haute résolution |
| Conséquence sur EW (si mauvais choix) | Sous-estimation de l'EW réelle (manque les ailes de pression). | Surestimation si le centre est trop rond. | Le meilleur compromis.       |

````python
# Simuler un spectre H-alpha (autour de 6563 Angstroms) avec bruit
x = np.linspace(6400, 6700, 500)

dash.clear_spectra()

# Profil gaussien centré sur H-alpha
y1 = 10 + 50 * np.exp(-0.5 * ((x - 6562.8)/5)**2)  + np.random.normal(0, 1, 500)
dash.load_spectrum(x, y1, label="Centre")

# Simuler un deuxième spectre décalé (effet Doppler)
y2 = 8 + 40 * np.exp(-0.5 * ((x - 6570)/5)**2) + np.random.normal(0, 1, 500)
dash.load_spectrum(x, y2, label="Bras Spiral (+300km/s)")
````
