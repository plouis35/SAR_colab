## une démonstration complète

````python
import numpy as np
import matplotlib.pyplot as plt
from astropy import units as u
from astropy.modeling import models
from specutils import Spectrum1D, SpectralRegion
from specutils.fitting import fit_lines
from specutils.analysis import centroid, fwhm, equivalent_width, snr, snr_derived

# --- 1. Simulation ---
nb_echantillons = 500
x = np.linspace(6400, 6700, nb_echantillons)
cont_reel = 1.0
bruit_sigma = 0.3
raie_obs = 6563.44

# Raie centrée à raie_obs, amplitude 4, largeur sigma 5
y = cont_reel + 4 * np.exp(-0.5 * ((x - raie_obs)/5)**2) + np.random.normal(0, bruit_sigma, nb_echantillons)

spec = Spectrum1D(flux=y*u.adu, spectral_axis=x*u.AA)
region_raie = SpectralRegion(6540.0 * u.AA, 6590.0 * u.AA)
region_cont = SpectralRegion(6400.0 * u.AA, 6500.0 * u.AA)

# --- 2. Calculs SANS ajustement ---
x_val = spec.spectral_axis.value
mask_indices = (x_val >= region_cont.lower.value) & (x_val <= region_cont.upper.value)

# On calcule la moyenne du flux dans ces indices
niveau_continuum_brut = np.mean(spec.flux[mask_indices])

# Normalisation : on divise par le niveau du continuum
flux_norm_val = spec.flux.value / niveau_continuum_brut.value
spec_norm = Spectrum1D(flux=flux_norm_val * u.dimensionless_unscaled, 
                       spectral_axis=spec.spectral_axis)

# Version sans plancher (continuum à 0) pour le centroïde et la FWHM
spec_norm_sans_cont = spec_norm - models.Const1D(1.0)(spec.spectral_axis) * u.dimensionless_unscaled

# Mesures
snr_val = snr_derived(spec, region_cont)
cent_brut = centroid(spec_norm_sans_cont, regions=region_raie)
fwhm_brut = fwhm(spec_norm_sans_cont, regions=region_raie)

# L'EW est calculée sur le spectre normalisé (continuum à 1.0)
ew_brut = equivalent_width(spec_norm, regions=region_raie)

# --- CALCUL DES ERREURS ---
err_cent_brut = fwhm_brut / snr_val
err_fwhm_brut = (2 * fwhm_brut) / snr_val
delta_lambda = np.mean(np.diff(x_val))
err_ew_brut = (1.5 / snr_val) * np.sqrt(delta_lambda * np.abs(ew_brut.value))

print(f"SNR mesuré : {snr_val:.2f}")
print("-" * 30 + " SANS AJUSTEMENT " + "-" * 30)
print(f"Centroïde : {cent_brut.value:.2f} +/- {err_cent_brut.value:.2f} Å")
print(f"FWHM      : {fwhm_brut.value:.2f} +/- {err_fwhm_brut.value:.2f} Å")
print(f"EW        : {ew_brut.value:.2f} +/- {err_ew_brut:.2f} Å")

# --- 3. Calculs AVEC ajustement ---
# On définit le modèle : Gaussienne + Continuum (Const1D)
g_init = models.Gaussian1D(amplitude=4*u.adu, mean=6563.4*u.AA, stddev=5*u.AA) + models.Const1D(1.0)
g_fit = fit_lines(spec, g_init)

# Extraction des résultats du fit
center_fit = g_fit[0].mean.value
fwhm_fit = g_fit[0].fwhm.value

# Incertitudes
err_center_fit = fwhm_fit / snr_val
err_fwhm_fit = (1.5 * fwhm_fit) / snr_val 

print("-" * 30 + " AVEC AJUSTEMENT " + "-" * 30)
print(f"Centroïde : {center_fit:.2f} +/- {err_center_fit:.2f} Å")
print(f"FWHM      : {fwhm_fit:.2f} +/- {err_fwhm_fit:.2f} Å")

# Extraction des paramètres pour le calcul de l'aire
amp_fit = g_fit[0].amplitude.value
sigma_fit = g_fit[0].stddev.value
continuum_fit = g_fit[1].amplitude.value

# 2. Calcul de l'EW du modèle
# L'aire d'une gaussienne est Amp * Sigma * sqrt(2*pi)
# L'EW est cette aire divisée par le niveau du continuum
aire_gaussienne = amp_fit * sigma_fit * np.sqrt(2 * np.pi)
ew_fit_val = aire_gaussienne / continuum_fit

# 3. Incertitude sur l'EW du fit (Formule de Cayrel)
# Le fit réduit l'incertitude par rapport au brut (facteur ~1.0 au lieu de 1.5)
err_ew_fit = (1.0 / snr_val) * np.sqrt(delta_lambda * np.abs(ew_fit_val))

print(f"EW        : {ew_fit_val:.2f} +/- {err_ew_fit:.2f} Å")

# --- 4. ANALYSE CINÉMATIQUE FINALE ---
c_kms = 299792.458
lambda_repos = 6562.80 

# Utilisation des valeurs du FIT (plus précises)
vr_fit = c_kms * (center_fit - lambda_repos) / lambda_repos
err_vr = c_kms * (0.32 / lambda_repos)

v_expansion = (c_kms * (fwhm_fit / lambda_repos)) / 2
err_v_exp = (c_kms * (0.48 / lambda_repos)) / 2 

print("\n" + "="*30 + " SYNTHÈSE ASTROPHYSIQUE " + "="*30)
print(f"Vitesse Radiale (Mouvement global) : {vr_fit:.2f} +/- {err_vr:.2f} km/s")
print(f"Vitesse d'Expansion (Gaz)          : {v_expansion:.2f} +/- {err_v_exp:.2f} km/s")
print("="*80)

# Plot the original spectrum and the fitted.
y_fit = g_fit(spec.spectral_axis)
plt.close('all')
plt.plot(x, y, label="brut")
plt.plot(x, y_fit, label="ajusté")
plt.title('raie simulée')
plt.grid(True)
plt.legend()
plt.show()

````

# analyse : que mesure-t-on ?

## les méthodes utilisée par les amateurs

| Mesure | Méthode | Implémentation | Incertitude |
| :--- | :--- | :--- | :--- |
| **Qualité des données** | **déviation standard** du flux ($\sigma_{flux}$) dans une région de continuum considérée comme "pure bruit" : $$\text{SNR} = \frac{1}{\sigma_{flux}}$$ |  `1 / astropy.stats.mad_std` | - |
| **Type spectral ou température**| **Ajustement** de la courbe de Planck : $$B_\lambda(T) = \frac{2hc^2}{\lambda^5} \cdot \frac{1}{e^{\frac{hc}{\lambda k_B T}} - 1}$$ | `popt, pcov = scipy.curve_fit` | $$\sigma_T = \sqrt{\text{pcov}[0][0]}$$ |
| **Vitesse de Rotation** ou **Élargissement Thermique** |**Largeur à mi-hauteur (FWHM)** d'un profil normalisé et ajusté. |  `specutils.fitting.fit_lines` | $\sigma_{\text{FWHM}} \approx \frac{\text{FWHM}}{\text{SNR}}$ |
| **Abondance Chimique** (Concentration des éléments) | **EW** (Largeur Équivalente) : **aire** d'un profil normalisée et ajustée. |  `specutils.analysis.equivalent_width` | $$\sigma_{\text{EW}} \approx \frac{1.5}{\text{SNR}} \sqrt{\delta\lambda \cdot \text{EW}}$$ |
| **Vitesse de l'Étoile** (Déplacement global de l'objet) | **VR / Centroïde** (Mono-raie) : Le **centre** ($x_0$) ajusté d'un profil nomalisé et ajusté. | `specutils.fitting.fit_lines` | $$\sigma_{v} \approx \frac{c \cdot \text{FWHM}}{2.35 \cdot \text{SNR} \cdot \sqrt{N_{\text{pts}}}}$$ |
| **Mouvement orbital** (de binaires) | **VR Global** (Multi-raies / CCF) : **moyenne** des vitesses radiales obtenues sur plusieurs raies individuelles. |  `corr, lag = specutils.correlation.template_correlate(spec_observed, spec_template)` | $$\sigma_{\text{tot}} = \frac{\text{StdDev}(\text{Liste VR})}{\sqrt{N_{\text{raies}}}}$$ |


Voici les trois principaux modèles utilisés en spectroscopie. Chacun de ces profils est une fonction mathématique qui modélise l'intensité du flux ($I$) en fonction de la longueur d'onde ($\lambda$) autour de la raie.


| Caractéristiques                      | Gaussienne                                                     | Lorentzienne                              | Voigt                        |
| :------------------------------------ | :------------------------------------------------------------- | :---------------------------------------- | :--------------------------- |
| Ailes                                 | Fines (tombent vite)                                           | Larges (tombent lentement)                | Larges (comme Lorentz)       |
| Zone du Sommet                        | Arrondie                                                       | Pointue                                   | Arrondie (dominée par Gauss) |
| Origine Physique                      | Doppler thermique (l'agitation aléatoire des atomes due à la chaleur) | Amortissement naturel (limite de la durée de vie des niveaux d'énergie) et élargissement par pression   | Combine agitation thermique et effets de pression/collision |
| Utilisation typique                   | VR, faible résolution                                          | Raies dominées par pression (H)           | EW précise, haute résolution |
| Conséquence sur EW (si mauvais choix) | Sous-estimation de l'EW réelle (manque les ailes de pression). | Surestimation si le centre est trop rond. | Le meilleur compromis.       |

## fonctions utiles

````python
import numpy as np
import matplotlib.pyplot as plt
from astropy.modeling import models
from astropy.stats import mad_std
from astropy import units as u
from specutils.spectra import SpectralRegion
from specutils.manipulation import extract_region

def get_snr_continuum(spectrum: Spectrum1D, region: SpectralRegion) -> float:
    """
    retourne le SNR en calculant la déviation standard du flux dans une région
    de continuum spécifiée, en utilisant la méthode robuste MAD_STD.
    Paramètres :
    - flux (Spectrum1D) : spectre à analyser.
    - region (SpectralRegion) :  zone du spectre en Ang
    """
    
    # récupère la tranche de flux (spectre normalisé, moyenne ~ 1)
    flux_continuum = extract_region(spectrum, region)

    if len(flux_continuum.spectral_axis) < 10:
        raise ValueError("Zone trop petite")
            
    # le bruit (sigma_flux) en utilisant la MAD
    sigma_flux = mad_std(flux_continuum.flux)
    
    # Signal_moyen = np.mean(flux_continuum) # si spectre non normalisé
    Signal_moyen = 1.0 

    # Calcul du SNR
    if sigma_flux > 0:
        SNR_local = Signal_moyen / sigma_flux
    else:
        SNR_local = np.inf
        
    return SNR_local
````

````python
region_continuum =  SpectralRegion(6400.0 * u.AA, 6500.0 * u.AA)

print(f"{get_snr_continuum(spec, region_continuum):.2f}")
````

````python
from specutils.analysis import equivalent_width

# Constantes
VITESSE_LUMIERE = 299792.458  # km/s

def get_equivalent_width(spectrum, region, SNR) -> (float, float):
    """
    retourne la largeur équivalente (EW)  et son incertitude (sigma_EW)
    """
    if SNR <= 0:
        raise ValueError("SNR < 0")

    # pas spectral
    delta_lambda = np.mean(np.diff(spectrum.spectral_axis)) # Pas spectral (dx)

    # ew
    ew_value = equivalent_width(spectrum, regions=region)
    print(f"{ew_value}")
    
    # sigma_ew (Formule Cayrel)
    sigma_ew = (1.5 / SNR) * np.sqrt(delta_lambda.value * np.abs(ew_value.value))

    return sigma_ew
````
